### Solution 1

```

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0, stratify=y)

scaler = StandardScaler().fit(X_train)
X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)

y_pred = knn.predict(X_test_scaled)
print(f"Test accuracy: {accuracy_score(y_test, y_pred):.3f}")
```

### Solution 2

```

scaler_full = StandardScaler().fit(X)
X_scaled_full = scaler_full.transform(X)

kmeans = KMeans(n_clusters=3, random_state=0, n_init=10)
cluster_labels = kmeans.fit_predict(X_scaled_full)

score = adjusted_rand_score(y, cluster_labels)
print(f"Adjusted Rand Index (cluster assignments vs. real cultivars): {score:.3f}")
print("(1.0 = perfect agreement, 0.0 = no better than random. This dataset's features happen to separate")
print(" the cultivars well enough that KMeans gets remarkably close, with zero label information used.)")
```

### Solution 3

```

print("KNN stores the entire training set:")
print("  knn._fit_X.shape:", knn._fit_X.shape)
print("  len(X_train):", len(X_train))
print()
print("KNN's 'model' IS the training data — every one of the", len(X_train), "training rows is retained")
print("and searched at prediction time. Double the training data, and KNN's storage AND prediction-time")
print("cost both roughly double too.")
print()
print("A LogisticRegression model, by contrast, would store exactly one weight per feature")
print(f"({X.shape[1]} numbers) plus one intercept — regardless of whether it was trained on")
print("142 rows (like here) or 142,000. That fixed-size-regardless-of-data-volume property is exactly")
print("the parametric/non-parametric line from the lesson.")
```
